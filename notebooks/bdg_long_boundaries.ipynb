# Longer BdG chains with smooth boundaries
Bulk parameters gs=gc=J=1, Delta=0.1, phi=pi/2, P=0. Joint taper: s_j=tanh(min(j,N+1-j)/N0), Delta_j=0.1 s_j^2, gs_j=s_j^2, gc_(j,j+1)=s_j s_(j+1). Uniform amplitude damping kappa/2. Compare fixed N0=8 with N0=N/8 and N0=N/5 at N=128,256,512.
This is the same prescribed quadratic adaptation as the previous study, not a nonlinear mean-field solution. See `../derivations/bdg_long_boundary_scaling.md`.

Thresholds use high-precision characteristic polynomials and full Routh root counts. Candidate real roots are accepted only after counting all roots on either side of the bracket. Both bracket counts are repeated with 100 additional decimal digits.
$$
\Gamma_N(\kappa)=(\kappa_c-\kappa)/2.
$$
Strict stability requires kappa above the threshold; the saved upper bracket provides the conservative numerical decision.
