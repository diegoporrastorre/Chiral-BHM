## Exact predictions
$$
\Gamma_{\rm OBC}=\Gamma_{\rm bulk}(0)=1-\kappa/2,\qquad \Gamma_{\rm PBC}=3-\kappa/2.
$$
At zero reference frequency the determinant winding is +1 for 2<kappa<6 and zero outside; it is undefined at 2 and 6. At kappa=2 the OBC propagator grows polynomially for N>1. The constant channel rotation removes boundary conversion exactly; topology is not the reason the bulk criterion succeeds here.
