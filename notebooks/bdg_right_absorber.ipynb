The reflection amplitude uses unit-norm bulk mode vectors, not flux normalization. In the unstable regime it is a formal frequency-domain scattering diagnostic. Stability is established separately by high-precision root counts.
$$
R_{\mathrm{required}}(u)=-\frac{v_{R,q}}{v_{L,q}}\left(\frac{z_L}{z_R}\right)^M,\qquad u=\kappa/2-i\omega.
$$
A large round-trip magnitude alone is not a stability proof: phase and subsystem poles matter. The exactly transparent benchmark is a frequency-dependent self-energy; it is not one of the tested static spatial profiles.
