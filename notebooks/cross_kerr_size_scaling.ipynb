# Size dependence of compensated intermediate states
Compare the trivial slice at $\phi=35^\circ$ with the topological slice at $130^\circ$, at $J=\kappa=1$, $\Delta=3$, $U=-0.0002$, $V=4U$. The uniform mean-field background and its drive-only compensation stay fixed as chain length changes. Stability means $\max\operatorname{Im}\omega<0$ for $i\dot z=Lz$. The invariant is the bulk winding at zero frequency.

The calculation follows the stable component adjoining the lower-density fold and compares fixed distinct intermediate roots. The exact fold is excluded from the intermediate-state claim. Decimal characteristic polynomials and Routh counts check the non-Hermitian spectra. This is a finite-size numerical study, not an infinite-chain theorem.

See `docs/uniform_topological_segment_stability.md` for findings and `derivations/cross_kerr_fold_window_scaling.md` for the drive-width derivation.

```python
from pathlib import Path
import os, sys, json
root = Path.cwd()
if not (root / 'src').is_dir():
    root = root.parent
assert (root / 'src' / 'cross_kerr_size_scaling.py').is_file()
os.chdir(root)
sys.path.insert(0, str(root / '.local-deps'))
sys.path.insert(0, str(root / 'src'))
from cross_kerr_size_scaling import main
# Reuses per-point JSON caches; an uncached run takes several minutes.
main()
```

```python
from IPython.display import Markdown, Image, display
result = json.loads((root / 'data/generated/cross_kerr_trivial/size_scaling/summary.json').read_text())
lines = ['| N | Trivial drive width | Topological drive width | Trivial growth | Topological growth |', '|---:|---:|---:|---:|---:|']
for row in result['rows']:
    widths = [(row[k]['drive_width_lower'] + row[k]['drive_width_upper']) / 2 for k in ['trivial', 'topological']]
    rates = [(row[k]['growth_lower'] + row[k]['growth_upper']) / 2 for k in ['fixed_trivial', 'fixed_topological']]
    lines.append('| ' + str(row['N']) + ' | ' + ' | '.join(f'{x:.7g}' for x in widths + rates) + ' |')
display(Markdown('\n'.join(lines)))
display(Image(filename=str(root / 'figures/cross_kerr_size_scaling.png')))
```

The growth columns use fixed density fractions $f=0.001$ (trivial) and $f=0.1$ (topological). Width zero means the stable component touching the fold is absent; it does not rule out disconnected stable intervals. Fractions, growth brackets, precisions, winding checks, and stationary residuals are stored in the JSON output.

A persistent spectral stability margin does not establish robustness against errors in the compensation drives or generic perturbations. The comparison changes hopping phase and hence the bulk operator; topology has not been isolated as an independent causal variable.
